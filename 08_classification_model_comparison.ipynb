## Classification Model Comparison

This notebook compares the classification models used to predict whether an engine is likely to fail soon.

The models are evaluated using the same validation dataset and cost-sensitive thresholding approach.

Models compared:
- Logistic Regression
- Random Forest
- SVM

```python
import pandas as pd

comparison = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Random Forest",
        "SVM"
    ],
    
    "Threshold": [
        0.60,
        0.45,
        0.35
    ],
    
    "Accuracy": [
        0.94,
        0.96,
        0.95
    ],
    
    "Precision": [
        0.75,
        0.80,
        0.79
    ],
    
    "Recall": [
        0.95,
        0.95,
        0.94
    ],
    
    "F1 Score": [
        0.83,
        0.87,
        0.86
    ],
    
    "False Positives": [
        201,
        146,
        153
    ],
    
    "False Negatives": [
        32,
        33,
        36
    ],
    
    "Total Cost": [
        361,
        311,
        333
    ]
})

comparison
```

## Model Performance Comparison

The following visualization compares the classification models using accuracy, precision, recall, and F1-score.

```python
import matplotlib.pyplot as plt

metrics = ["Accuracy", "Precision", "Recall", "F1 Score"]

comparison.set_index("Model")[metrics].plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Classification Model Performance Comparison")
plt.ylabel("Score")
plt.ylim(0, 1)
plt.xticks(rotation=0)
plt.legend(title="Metrics")
plt.grid(axis="y", alpha=0.3)

plt.show()
```

## Observation

The three classification models show similar recall for the failure-soon class. Random Forest achieved the highest accuracy and F1-score among the
three models and also resulted in the lowest total cost after cost-sensitive threshold optimization.

The optimized thresholds were different for each model, showing that the default probability threshold of 0.50 is not necessarily optimal when false positives and false negatives have different costs.
